```python
# Calculate average price by car make
df.groupby('make')['price'].mean().sort_values(ascending=False)
```

```python
# Calculate average mileage by fuel type
df.groupby('fuel_type')['mileage'].mean()
```

```python
# Calculate average price by car make
plt.figure(figsize=(8,5))

plt.hist(df['price'], bins=30)

plt.xlabel("Price")
plt.ylabel("Number of Cars")
plt.title("Distribution of Car Prices")

plt.show()
```

```python
# Create a bar chart of the top 10 car makes
plt.figure(figsize=(10,6))

df['make'].value_counts().head(10).plot(kind='bar')

plt.xlabel("Car Make")
plt.ylabel("Number of Cars")
plt.title("Top 10 Car Makes")

plt.xticks(rotation=45)

plt.show()
```

```python
# Create a scatter plot of price versus mileage
plt.figure(figsize=(8,5))

plt.scatter(df['mileage'], df['price'])

plt.xlabel("Mileage")
plt.ylabel("Price")
plt.title("Price vs Mileage")

plt.show()
```

```python
# Create a correlation heatmap for numerical variables
corr = df[['year', 'price', 'engine_size', 'horsepower', 'mileage']].corr()

plt.figure(figsize=(10,6))

sns.heatmap(corr, annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")

plt.show()
```

```python
# Create a boxplot of car prices
plt.figure(figsize=(8,5))

sns.boxplot(x=df['price'])

plt.xlabel("Price")
plt.title("Boxplot of Car Prices")

plt.show()
```

```python
# Create a count plot of cars by fuel type
plt.figure(figsize=(8,5))

sns.countplot(data=df, x='fuel_type')

plt.xlabel("Fuel Type")
plt.ylabel("Number of Cars")
plt.title("Distribution of Cars by Fuel Type")

plt.xticks(rotation=45)

plt.show()
```

```python
# Create a histogram of horsepower
plt.figure(figsize=(8,5))

plt.hist(df['horsepower'], bins=30)

plt.xlabel("Horsepower")
plt.ylabel("Number of Cars")
plt.title("Distribution of Horsepower")

plt.show()
```

```python
# Detect outliers in car prices using the IQR method
Q1 = df['price'].quantile(0.25)
Q3 = df['price'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['price'] < lower) | (df['price'] > upper)]

print("Number of price outliers:", len(outliers))
```

```python
# Detect outliers in car prices using the IQR method
Q1 = df['mileage'].quantile(0.25)
Q3 = df['mileage'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['mileage'] < lower) | (df['mileage'] > upper)]

print("Number of mileage outliers:", len(outliers))
```
